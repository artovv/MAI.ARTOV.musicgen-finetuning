```python
from transformers import AutoFeatureExtractor
from datasets import Audio

instance_prompt = "my_style"


audio_encoder_feature_extractor = AutoFeatureExtractor.from_pretrained(
    model.config.audio_encoder._name_or_path,
)


dataset_sampling_rate = dataset[0]["audio"]["sampling_rate"]

if dataset_sampling_rate != audio_encoder_feature_extractor.sampling_rate:
    dataset = dataset.cast_column(
        "audio",
        Audio(
            sampling_rate=audio_encoder_feature_extractor.sampling_rate
        ),
    )



def prepare_audio_features(batch):
    metadata = batch["metadata"]
    metadata = f"{instance_prompt}, {metadata}"
    batch["input_ids"] = processor.tokenizer(metadata)["input_ids"]

    # load audio
    target_sample = batch["audio"]
    labels = audio_encoder_feature_extractor(
        target_sample["array"], sampling_rate=target_sample["sampling_rate"]
    )
    batch["labels"] = labels["input_values"]

    batch["target_length"] = len(target_sample["array"].squeeze())
    return batch


dataset = dataset.map(
    prepare_audio_features,
    remove_columns=dataset.column_names,
    num_proc=2,
    desc="preprocess datasets",
)
```

```python
import torch

audio_decoder = model.audio_encoder
num_codebooks = model.decoder.config.num_codebooks
audio_encoder_pad_token_id = model.config.decoder.pad_token_id

pad_labels = torch.ones((1, 1, num_codebooks, 1)) * audio_encoder_pad_token_id

if torch.cuda.device_count() == 1:
    audio_decoder.to("cuda")

def apply_audio_decoder(batch):

    with torch.no_grad():
        labels = audio_decoder.encode(
            torch.tensor(batch["labels"]).to(audio_decoder.device)
        )["audio_codes"]

    labels = torch.cat(
        [pad_labels.to(labels.device).to(labels.dtype), labels], dim=-1
    )

    labels, delay_pattern_mask = model.decoder.build_delay_pattern_mask(
        labels.squeeze(0),
        audio_encoder_pad_token_id,
        labels.shape[-1] + num_codebooks,
    )

    labels = model.decoder.apply_delay_pattern_mask(labels, delay_pattern_mask)

    batch["labels"] = labels[:, 1:].cpu()
    return batch

dataset = dataset.map(
    apply_audio_decoder,
    num_proc=1,
    desc="Apply encodec",
)
```

```python
from transformers import Seq2SeqTrainer, Seq2SeqTrainingArguments
from dataclasses import dataclass

class MusicgenTrainer(Seq2SeqTrainer):
    def _pad_tensors_to_max_len(self, tensor, max_length):
        if self.tokenizer is not None and hasattr(self.tokenizer, "pad_token_id"):
            pad_token_id = (
                self.tokenizer.pad_token_id
                if self.tokenizer.pad_token_id is not None
                else self.tokenizer.eos_token_id
            )
        else:
            if self.model.config.pad_token_id is not None:
                pad_token_id = self.model.config.pad_token_id
            else:
                raise ValueError(
                    "Pad_token_id must be set in the configuration of the model, in order to pad tensors"
                )

        padded_tensor = pad_token_id * torch.ones(
            (tensor.shape[0], max_length, tensor.shape[2]),
            dtype=tensor.dtype,
            device=tensor.device,
        )
        length = min(max_length, tensor.shape[1])
        padded_tensor[:, :length] = tensor[:, :length]
        return padded_tensor


@dataclass
class DataCollatorMusicGenWithPadding:
    processor: AutoProcessor

    def __call__(
        self, features
    ):
        labels = [
            torch.tensor(feature["labels"]).transpose(0, 1) for feature in features
        ]

        labels = torch.nn.utils.rnn.pad_sequence(
            labels, batch_first=True, padding_value=-100
        )

        input_ids = [{"input_ids": feature["input_ids"]} for feature in features]
        input_ids = self.processor.tokenizer.pad(input_ids, return_tensors="pt")

        batch = {"labels": labels, **input_ids}

        return batch

data_collator = DataCollatorMusicGenWithPadding(
    processor=processor,
)

```

```python
from transformers import (
    AutoProcessor,
    AutoModelForTextToWaveform,
)

processor = AutoProcessor.from_pretrained("facebook/musicgen-melody")
model = AutoModelForTextToWaveform.from_pretrained("facebook/musicgen-melody")

model.freeze_text_encoder()
model.freeze_audio_encoder()
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

import torch
import torch.nn as nn
import torch.nn.functional as F

class FourierFTWrapper(nn.Module):
    def __init__(self, base_layer: nn.Linear, n: int = 100, alpha: float = 300.0,
                 freq_bias: str = "middle", bandwidth: float = 0.2):
        super().__init__()
        self.base = base_layer
        self.alpha = alpha
        self.d1, self.d2 = base_layer.weight.shape
        self.n = min(n, self.d1 * self.d2)
        # Инициализация спектральной матрицы с bias
        E = self._sample_entries_with_frequency_bias(freq_bias, bandwidth)
        self.register_buffer("E", E)
        # Обучаемые спектральные коэффициенты
        self.delta_W = nn.Parameter(torch.randn(self.n))

        # Заморозка base-слоя
        for param in self.base.parameters():
            param.requires_grad = False

    def _sample_entries_with_frequency_bias(self, freq_bias, bandwidth):
        bias_map = {
            "low": 0.1,
            "middle": 0.5,
            "high": 0.9,
            "none": None  # равномерное распределение
        }
        fc = bias_map.get(freq_bias, 0.5)
        d1, d2 = self.d1, self.d2

        # Координаты в частотной области
        u = torch.arange(d1).unsqueeze(1).expand(d1, d2)
        v = torch.arange(d2).unsqueeze(0).expand(d1, d2)
        if fc is None:
            prob = torch.ones(d1, d2)
        else:
            # Центр в частотной матрице
            center_u = d1 * fc
            center_v = d2 * fc
            D = ((u - center_u) ** 2 + (v - center_v) ** 2).sqrt()
            DW = bandwidth * max(d1, d2)
            prob = torch.exp(-((D - fc * max(d1, d2)) / DW) ** 2)
        # Нормализация
        prob = prob / prob.sum()
        prob_flat = prob.flatten()

        # Сэмплирование n индексов
        indices = torch.multinomial(prob_flat, self.n, replacement=False)
        E0 = indices // d2
        E1 = indices % d2
        return torch.stack([E0, E1], dim=0)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        h = self.base(x)

        device = x.device
        delta_W = self.delta_W.to(device)
        E0 = self.E[0].to(device)
        E1 = self.E[1].to(device)

        F = torch.zeros(self.d1, self.d2, dtype=torch.float32, device=device)
        F[E0, E1] = delta_W

        delta_W_matrix = torch.fft.ifft2(F).real * self.alpha

        if x.shape[-1] != delta_W_matrix.shape[0]:
            raise ValueError(f"[FourierFT] Shape mismatch: x={x.shape}, delta={delta_W_matrix.shape}")

        h += torch.einsum("...d,dc->...c", x, delta_W_matrix)
        return h


def test_fourierft_gradient():
    torch.manual_seed(42)

    base_layer = nn.Linear(64, 64, bias=False)
    fourier_layer = FourierFTWrapper(base_layer, n=10, alpha=300.0)

    optimizer = torch.optim.AdamW([p for p in fourier_layer.parameters() if p.requires_grad], lr=1e-3)

    x = torch.randn(2, 64, requires_grad=True)
    target = torch.randn(2, 64)


    y = fourier_layer(x)
    loss = F.mse_loss(y, target)
    loss.backward()

    print(f"Loss: {loss.item()}")
    print(f"delta_W requires_grad: {fourier_layer.delta_W.requires_grad}")
    print(f"delta_W.grad is None? {fourier_layer.delta_W.grad is None}")
    print(f"delta_W.grad norm: {fourier_layer.delta_W.grad.norm().item():.6f}" if fourier_layer.delta_W.grad is not None else "No grad")

test_fourierft_gradient()
```

```python
def inject_fourierft(model, n=1000, alpha=300.0, freq_bias="middle", bandwidth=0.2):
    total = 0
    for i, layer in enumerate(model.decoder.model.decoder.layers):
        for name in ["v_proj", "q_proj"]:
            target = getattr(layer.self_attn, name)
            if isinstance(target, nn.Linear):
                ft = FourierFTWrapper(
                    target,
                    n=n,
                    alpha=alpha,
                    freq_bias=freq_bias,  # "low", "middle", "high", "none"
                    bandwidth=bandwidth
                )
                setattr(layer.self_attn, name, ft)
                total += 1
    print(f"Всего вставлено адаптеров: {total}")
```

```python
def freeze_all_except_fourier(model):
    frozen = 0
    trainable = 0
    for name, param in model.named_parameters():
        if "delta_W" in name:
            param.requires_grad = True
            trainable += param.numel()
        else:
            param.requires_grad = False
            frozen += param.numel()
    print(f"Заморожено параметров: {frozen:,}, обучаемо (Fourier): {trainable:,}")
```

```python
from msclap import CLAP
import torch
import torch.nn.functional as F
from torch.utils.data import DataLoader
from torch.amp import autocast, GradScaler
from copy import deepcopy
import optuna
from optuna.samplers import TPESampler
import random, gc, os, tempfile
import torchaudio
import torchaudio.transforms as T
import numpy as np
import pandas as pd

clap_model = CLAP(version="2023", use_cuda=True)
NUM_BATCHES = 3
NUM_EVAL_SAMPLES = 4
CLAP_WEIGHT = 0.7
LOSS_WEIGHT = 0.3

trial_logs = []

def clear_cuda_memory():
    torch.cuda.empty_cache()
    torch.cuda.ipc_collect()
    gc.collect()

def objective(trial):
    clear_cuda_memory()
    try:
        n = trial.suggest_int("n", 500, 5000, step=500)
        alpha = trial.suggest_float("alpha", 50.0, 400.0)
        freq_bias = trial.suggest_categorical("freq_bias", ["none", "low", "middle", "high"])
        bandwidth = trial.suggest_float("bandwidth", 0.1, 0.3)
        lr = trial.suggest_float("lr", 1e-4, 6e-2, log=True)
        grad_accum = trial.suggest_categorical("grad_accum", [1, 8])
        guidance_scale = trial.suggest_float("guidance_scale", 1.0, 5.0)

        model = deepcopy(base_model).cuda().train()
        inject_fourierft(model, n=n, alpha=alpha, freq_bias=freq_bias, bandwidth=bandwidth)
        freeze_all_except_fourier(model)

        optimizer = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=lr)
        scaler = GradScaler()

        dataloader = DataLoader(dataset, batch_size=1, shuffle=True, collate_fn=data_collator)
        dataloader_iter = iter(dataloader)
        total_loss = 0.0

        for _ in range(NUM_BATCHES):
            batch = next(dataloader_iter)
            for key in batch:
                if isinstance(batch[key], torch.Tensor):
                    batch[key] = batch[key].to(model.device)
            with autocast("cuda"):
                outputs = model(**batch)
                loss = outputs.loss / grad_accum
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad()
            total_loss += loss.item()

        avg_loss = total_loss / NUM_BATCHES

        sim_values = []
        for _ in range(NUM_EVAL_SAMPLES):
            idx = random.randint(0, len(dataset_original) - 1)
            audio_array = dataset_original[idx]["audio"]["array"].astype(np.float32)
            sr = dataset_original[idx]["audio"]["sampling_rate"]
            audio_array /= (np.abs(audio_array).max() + 1e-9)
            prompt = f"my_style, {dataset_original[idx].get('metadata', 'pop, 120 bpm, guitar')}"

            if sr != 48000:
                resampler = T.Resample(sr, 48000)
                audio_tensor = torch.tensor(audio_array).unsqueeze(0)
                audio_tensor = resampler(audio_tensor).squeeze(0)
            else:
                audio_tensor = torch.tensor(audio_array)
            audio_tensor /= (audio_tensor.abs().max() + 1e-9)

            with tempfile.TemporaryDirectory() as tempdir:
                path = os.path.join(tempdir, "temp.wav")
                torchaudio.save(path, audio_tensor.unsqueeze(0), sample_rate=48000)
                with torch.no_grad():
                    audio_embed = clap_model.get_audio_embeddings([path])
                    text_embed = clap_model.get_text_embeddings([prompt])
                    sim = F.cosine_similarity(text_embed, audio_embed).item()
                    sim_values.append(sim)

        mean_sim = np.mean(sim_values)

        if mean_sim < 0.1:
            raise optuna.exceptions.TrialPruned()

        final_score = CLAP_WEIGHT * mean_sim + LOSS_WEIGHT * (-avg_loss)

        print(f"SIM={mean_sim:.4f} | LOSS={avg_loss:.4f} → Score={final_score:.4f}")
        print(f"n={n}, α={alpha:.1f}, lr={lr:.1e}, accum={grad_accum}, scale={guidance_scale:.2f}, bias={freq_bias}")

        trial.set_user_attr("clap", mean_sim)
        trial.set_user_attr("loss", avg_loss)

        trial_logs.append({
            "score": final_score,
            "n": n,
            "alpha": alpha,
            "lr": lr,
            "freq_bias": freq_bias,
            "guidance_scale": guidance_scale,
            "accum": grad_accum,
            "loss": avg_loss,
            "sim": mean_sim
        })

        del model, optimizer, scaler, batch, outputs
        clear_cuda_memory()
        return final_score

    except Exception as e:
        print(f"Trial failed: {e}")
        clear_cuda_memory()
        return -1.0
```

```python
from torch.cuda.amp import GradScaler
import os


sampler = TPESampler(multivariate=True)
study = optuna.create_study(direction="maximize", sampler=sampler)
study.optimize(objective, n_trials=100, timeout=None)

trial_logs_df = pd.DataFrame(trial_logs)
trial_logs_df = trial_logs_df.astype({
    "score": float,
    "n": int,
    "alpha": float,
    "lr": float,
    "guidance_scale": float,
    "accum": int,
    "loss": float,
    "sim": float
})

print("\nBest Trial:")
best = study.best_trial
for k, v in best.params.items():
    print(f"{k}: {v:.1e}" if k == "lr" else f"🔧 {k}: {v}")
print(f"Best Score: {best.value:.4f}")

best_clap = best.user_attrs.get("clap", None)
best_loss = best.user_attrs.get("loss", None)

if best_clap is not None and best_loss is not None:
    print(f"Best CLAP Similarity: {best_clap:.4f}")
    print(f"Best CrossEntropy Loss: {best_loss:.4f}")
else:
    print("Best trial does not contain clap/loss user_attrs.")
```

```python
import torch
import gc
import os
import soundfile as sf
from tqdm import tqdm
from torch.utils.data import DataLoader
from torch import autocast, GradScaler
from IPython.display import Audio, display
from torch import autocast

num_epochs = 30
batch_size = 2
lr = 1.7e-2
grad_accum_steps = 8
prompt = "my_style, acousticguitar, Pop, Dance-Pop, groovy"
guidance_scale = 4.6
gen_dir = "./intermediate_audio"
os.makedirs(gen_dir, exist_ok=True)
losses = []

scaler = GradScaler()
model.train().cuda()
compiled_model = model


dataloader = DataLoader(dataset, batch_size=batch_size, shuffle=True, collate_fn=data_collator, num_workers=8, pin_memory=True, persistent_workers=True)
optimizer = torch.optim.AdamW(
    [p for p in compiled_model.parameters() if p.requires_grad],
    lr=lr,
    betas=(0.9, 0.99),
    weight_decay=0.1,
)


sample_batch = next(iter(dataloader))
sample_batch = {k: v.to(compiled_model.device) for k, v in sample_batch.items()}
compiled_model.eval()
with torch.no_grad():
    with autocast("cuda"):
        outputs_before = compiled_model(**sample_batch).logits.detach().cpu()


for epoch in range(num_epochs):
    print(f"\nЭпоха {epoch + 1}/{num_epochs}")
    total_loss = 0.0
    compiled_model.train()
    gc.collect()
    torch.cuda.empty_cache()

    for step, batch in enumerate(tqdm(dataloader)):
        batch = {k: v.to(compiled_model.device) for k, v in batch.items()}

        with autocast(dtype=torch.float16, device_type="cuda"):
            outputs = compiled_model(**batch)
            loss = outputs.loss / grad_accum_steps

        scaler.scale(loss).backward()

        if (step + 1) % grad_accum_steps == 0:
            scaler.step(optimizer)
            scaler.update()
            optimizer.zero_grad(set_to_none=True)

        total_loss += loss.item() * grad_accum_steps

        if step % 10 == 0:
            print(f"Step {step} | Loss: {loss.item() * grad_accum_steps:.4f}")

    print(f"Средний loss: {total_loss / len(dataloader):.6f}")
    losses.append(total_loss / len(dataloader))
```

```python
import torch
import os
import gc
import io
import numpy as np
import soundfile as sf
import torch.nn.functional as F
from tqdm import tqdm
from torch.utils.data import DataLoader
from torch.amp import GradScaler, autocast
from msclap import CLAP

num_epochs = 30
batch_size = 2
lr = 3e-2
grad_accum_steps = 8
prompt = "my_style, Electronic, Dance-pop, acousticguitar, groovy"
guidance_scale = 4.6
CLAP_WEIGHT = 0.9
CE_WEIGHT = 0.1
margin = 0.15
num_candidates = 3
gen_dir = "./intermediate_audio"
os.makedirs(gen_dir, exist_ok=True)

scaler = GradScaler()
model.train().cuda()
compiled_model = model
clap_model = CLAP(version="2023", use_cuda=True)

dataloader = DataLoader(
    dataset, batch_size=batch_size, shuffle=True,
    collate_fn=data_collator, num_workers=4, pin_memory=True
)
optimizer = torch.optim.AdamW(
    [p for p in compiled_model.parameters() if p.requires_grad],
    lr=lr, betas=(0.9, 0.99), weight_decay=0.1
)

sample_batch = next(iter(dataloader))
sample_batch = {k: v.to(compiled_model.device) for k, v in sample_batch.items()}
compiled_model.eval()
with torch.no_grad(), autocast("cuda"):
    outputs_before = compiled_model(**sample_batch).logits.detach().cpu()

clap_avg_per_epoch = []
for epoch in range(num_epochs):
    print(f"\nЭпоха {epoch + 1}/{num_epochs}")
    total_loss, total_sim = 0.0, 0.0
    compiled_model.train()
    gc.collect()
    torch.cuda.empty_cache()

    for step, batch in enumerate(tqdm(dataloader)):
        batch = {k: v.to(compiled_model.device) for k, v in batch.items()}

        with autocast(dtype=torch.float16, device_type="cuda"):
            outputs = compiled_model(**batch)
            base_loss = outputs.loss / grad_accum_steps

        compiled_model.eval()
        audio_candidates = []
        with torch.no_grad():
            inputs = processor(text=[prompt], return_tensors="pt", padding=True).to(compiled_model.device)
            for temp in np.linspace(0.7, 1.5, num_candidates):
                audio = compiled_model.generate(
                    **inputs, do_sample=True, temperature=temp,
                    guidance_scale=guidance_scale, max_new_tokens=256
                )
                audio_np = audio.squeeze().detach().cpu().float().numpy()
                audio_np /= (np.abs(audio_np).max() + 1e-9)
                audio_bytes = io.BytesIO()
                sf.write(audio_bytes, audio_np, 32000, format="WAV")
                audio_bytes.seek(0)
                audio_candidates.append(audio_bytes)
        compiled_model.train()

        # CLAP-сходства
        similarities = []
        for audio_bytes in audio_candidates:
            try:
                with torch.no_grad():
                    audio_embed = clap_model.get_audio_embeddings([audio_bytes])
                    text_embed = clap_model.get_text_embeddings([prompt])

                    if audio_embed.ndim == 1:
                        audio_embed = audio_embed.unsqueeze(0)
                    if text_embed.ndim == 1:
                        text_embed = text_embed.unsqueeze(0)

                    sim = F.cosine_similarity(
                        F.normalize(audio_embed, dim=1),
                        F.normalize(text_embed, dim=1),
                        dim=1
                    ).item()
                    similarities.append(sim)
            except Exception as e:
                print(f"Ошибка CLAP: {e}")
                similarities.append(0.0)

        # Выбор winner и loser
        best_idx = int(np.argmax(similarities))
        worst_idx = int(np.argmin(similarities))
        winner_sim = similarities[best_idx]
        loser_sim = similarities[worst_idx]
        winner_audio_bytes = audio_candidates[best_idx]

        # Loss
        sim_diff = torch.tensor(margin - (winner_sim - loser_sim), device=compiled_model.device)
        rank_loss = F.softplus(sim_diff)
        final_loss = CLAP_WEIGHT * rank_loss + CE_WEIGHT * base_loss

        if not torch.isnan(final_loss):
            scaler.scale(final_loss).backward()
            if (step + 1) % grad_accum_steps == 0:
                scaler.step(optimizer)
                scaler.update()
                optimizer.zero_grad(set_to_none=True)

        total_loss += base_loss.item() * grad_accum_steps
        total_sim += winner_sim

        if step % 10 == 0:
            print(f"Step {step} | Rank Loss: {rank_loss:.4f} | Win: {winner_sim:.4f} | Lose: {loser_sim:.4f}")

    avg_clap = total_sim / len(dataloader)
    print(f"Средний CE Loss: {total_loss / len(dataloader):.6f}")
    print(f"Средний CLAP (winner): {total_sim / len(dataloader):.4f}")

    clap_avg_per_epoch.append(avg_clap)
    gc.collect()
    torch.cuda.empty_cache()
```
